```python
### configuration
import uuid
from pyspark.sql import functions as F
catalog = "taxi"
SOURCE_PATH = f"/Volumes/{catalog}/raw/landing/yellow"
CHECK_POINT = f"Volumes/{catalog}/raw/landing/_checkpoints/bronze_yellow_trips"
SCHEMA_PATH = f"/Volumes/{catalog}/raw/landing/_schemas/bronze_yellow_trips"
TARGET_TABLE = f"{catalog}.bronze.yellow_trips"
run_id = str(uuid.uuid4())
print("run id: ",run_id)
```

```python
### read with auto loader
raw_stream = spark.readStream.format("cloudFiles")\
    .option("cloudFiles.format","parquet")\
    .option("cloudFiles.schemaLocation",SCHEMA_PATH)\
    .load(SOURCE_PATH)


# readStream + cloudFiles is Auto Loader, which watches a folder and picks up only files it hasn't processed.
# cloudFiles.format tells it the files are Parquet.
# schemaLocation is where it keeps the inferred schema.
# .load(SOURCE_PATH) points it at your yellow/ folder.
```

```python
### add ingestion meta data
bronze_df= raw_stream.select("*",
                             F.col("_metadata.file_patch").alias("_source_file"),
                             F.col("_metadata.file_modification_time").alias("_file_modified_time"),
                             F.current_timestamp().alias("_ingest_time"),
                             F.lit(run_id).alias("_batch_id"),
                             )
```

<!-- uuid generates a unique ID, and functions (as F) gives you Spark column operations used later.

All locations are defined once at the top, so a rename means changing one line rather than hunting through the notebook.

SOURCE_PATH is where new files arrive.
CHECKPOINT is where Spark records which files it has already processed. This is the memory that makes the pipeline incremental.
SCHEMA_PATH is where Auto Loader stores the schema it inferred, so it can detect changes between files.
TARGET_TABLE is the Delta table you're creating.

run_id = str(uuid.uuid4())

This creates a unique ID for this run. It's stamped on every row loaded in this run (as _batch_id) and will later be used in the run log, so you can trace any row back to the run that loaded it. -->
